```python
# Importation des bibliothèques nécessaires.
import pandas as pd

from urllib.request import urlopen

from bs4 import BeautifulSoup as bs
```

```python
# Chargement des données depuis un fichier CSV.

# URL à scraper (exemple pour SFDS)
url = "https://www.sfds.asso.fr/fr/n/506-consulter_les_offres_demploi/?jedu=MASTER&jcon=&jp=3"

# Lire le contenu HTML de la page
page = urlopen(url)
```

```python
# page.read()
```

```python
html = bs(page, "html.parser")
```

```python
allJobs = html.find_all('div', class_='job')
# jobLabel 
```

```python
allJobs[0]
```

```python
for job in allJobs: 
    jobLabel = job.find('td', class_='jobLabel').get_text(strip=True)
    jobPublished = job.find('td', class_='jobPublished').get_text(strip=True)
    jobResume = job.find('td', class_='jobResume').get_text(strip=True)
    jobDetails = job.find('div', class_='jobDetails').get_text(strip=True)
    jobListLabel = job.find('td', class_='jobListLabel').get_text(strip=True)
    jobListValue = job.find('td', class_='jobListValue').get_text(strip=True)
    print(f"Intitulé du poste: {jobLabel}")
    print(f"Localisation: {jobPublished}")
    print(f"Résume: {jobResume}")
    print(f"Résume: {jobListLabel}")
    print(f"Résume: {jobListValue}")
    print("-" * 40) 
```
